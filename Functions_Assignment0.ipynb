```python
def sum_all(*numbers):
    return sum(numbers)

print(sum_all(1, 2, 3, 4))
```

```python
def print_info(**info):
    for key, value in info.items():
        print(f"{key}: {value}")

print_info(name="Alice", age=30)
```

ans-

*   **Iterables** are objects that can be looped over (iterated), like lists, tuples, strings, and dictionaries. They implement the __iter__() method.

*   **Iterators** are objects that actually perform the iteration. They implement both __iter__() and __next__() methods.

**Key Differences:**

| Feature        | Iterable               | Iterator               |
|----------------|------------------------|------------------------|
| Method         | Has `__iter__()`       | Has `__iter__()` and `__next__()` |
| Purpose        | Can be looped over     | Produces next item in iteration |
| Example types  | list, tuple, str       | Object from `iter()`   |



```python
# Iterable
my_list = [1, 2, 3]

# Create an iterator from the iterable
my_iterator = iter(my_list)

# Use the iterator to manually access elements
print(next(my_iterator))  # Output: 1
print(next(my_iterator))  # Output: 2
print(next(my_iterator))  # Output: 3
```

Trying next(my_iterator) again would raise StopIteration because there are no more elements.

# 6) **Explain the concept of generators in Python and how they are defined**.

ans- **Generators in Python:**
A generator is a type of function that returns an iterator, but instead of using return, it uses the yield keyword to produce a series of values one at a time, pausing after each yield and resuming from the same point when called again. This makes generators memory-efficient and suitable for processing large datasets or infinite sequences.

**How to Define a Generator:**
A generator is defined like a normal function, but uses yield to return values.

   Example:

```python
def even_numbers(limit):
    num = 0
    while num <= limit:
        yield num
        num += 2
```

**Using the Generator:**

```python
gen = even_numbers(6)

for n in gen:
    print(n)
```

**Key Points:**
Generators do not store all values in memory.

They are useful for iterating over large or infinite sequences.

Once exhausted, a generator cannot be reused unless recreated.

# 7. **What are the advantages of using generators over regular functions?**

ans-

1.  **Memory Efficiency:**
  Generators yield items one at a time, rather than storing the entire result in memory. This is ideal for working with large datasets or streams of data.

2. ** Lazy Evaluation:**
  Generators compute values on the fly and only when requested, which can lead to performance improvements and faster startup times.

3. **Improved Performance:**
   Since values are produced only when needed, generators can be faster and more responsive in certain scenarios, especially when not all results are required.

 4. **Simpler Code for Iteration:  **
   Generators eliminate the need for managing loop states manually, making the code cleaner and easier to maintain.

 5. **Can Model Infinite Sequences:**
  Unlike regular functions that return a complete result, generators can represent infinite sequences, like an endless stream of sensor data.

    Example Comparison:

      **Regular function:**



```python
def get_squares(n):
    result = []
    for i in range(n):
        result.append(i * i)
    return result
```

**Generator version:**

```python
def get_squares(n):
    for i in range(n):
        yield i * i
```

**Conclusion:**
 Use generators when working with large datasets, streams, or when you want to reduce memory usage.

# 8.What is a lambda function in Python and when is it typically used?

ans -** Lambda Function in Python:**
A lambda function is a small, anonymous function defined using the lambda keyword. It can have any number of arguments, but only one expression, which is returned automatically.
 Syntax:

```python
lambda arguments: expression
```

**When It's Typically Used:**


*   When you need a short, throwaway function for a short period of time.

*  Often used with functions like map(), filter(), and sorted() where defining a full function is unnecessary.

     Example:



```python
# Lambda function to double a number
double = lambda x: x * 2

print(double(5))  # Output: 10
```

**Used with map():**

```python
numbers = [1, 2, 3, 4]
doubled = list(map(lambda x: x * 2, numbers))
print(doubled)  # Output: [2, 4, 6, 8]
```

Lambda functions help keep your code concise and readable when a full def block would be overkill.

# **9. Explain the purpose and usage of the map() function in Python**?

ans- **Purpose of map() in Python:**
The map() function is used to apply a function to every item in an iterable (like a list or tuple) and returns a new map object (which is an iterator) with the results.

   **Syntax:**

```python
map(function, iterable)
```

**function**: A function to apply to each element.

**iterable:** A sequence (like a list) whose elements will be processed

   **Example:**

```python
# Function to square a number
def square(x):
    return x * x

numbers = [1, 2, 3, 4, 5]

# Apply square() to each number in the list
squared_numbers = list(map(square, numbers))

print(squared_numbers)  # Output: [1, 4, 9, 16, 25]
```

**With a Lambda Function:**

```python
numbers = [1, 2, 3]
result = list(map(lambda x: x + 10, numbers))
print(result)  # Output: [11, 12, 13]
```

**Key Benefits:**


*  Makes code cleaner and shorter
* Works well with lambda functions
*  Avoids explicit loops






# **10. What is the difference between map(), reduce(), and filter() functions in Python?**

ans- **Difference Between map(), reduce(), and filter() in Python:**
these are all functional programming tools used to process iterables, but they serve different purposes:

 | Function | Purpose | Returns | Use Case |
|----------|---------|---------|----------|
| `map()` | Applies a function to **each item** | Map object (iterator) | Transforming data |
| `filter()` | Filters items based on a **condition** | Filter object (iterator) | Selecting data |
| `reduce()` | **Reduces** iterable to a single value using a function | Final result (not iterable) | Aggregating data |

reduce() must be imported from functools in Python 3.


**Examples:**
1. map() – Apply a function to each item:



```python
numbers = [1, 2, 3, 4]
squares = list(map(lambda x: x**2, numbers))
print(squares)  # Output: [1, 4, 9, 16]
```

2. filter() – Keep items that match a condition:

```python
numbers = [1, 2, 3, 4, 5]
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens)  # Output: [2, 4]
```

3. reduce() – Reduce to a single value:

```python
from functools import reduce

numbers = [1, 2, 3, 4]
total = reduce(lambda x, y: x + y, numbers)
print(total)  # Output: 10
```

**Summary:**


*   Use **map()** to transform each element.
*   Use **filter()** to select specific elements.
*  Use **reduce()** to combine elements into one.
